```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import random
from datetime import datetime, timedelta

%matplotlib inline
```

```python
def generate_weather(city="Pune"):
    base_temp = 30

    data = []
    time = datetime.now()

    for i in range(24):  # 24 hours data
        temp = base_temp + random.randint(-6, 8)
        humidity = random.randint(40, 95)

        data.append({
            "datetime": (time + timedelta(hours=i)).strftime("%Y-%m-%d %H:%M"),
            "temp": temp,
            "humidity": humidity,
            "city": city
        })

    return pd.DataFrame(data)

df = generate_weather("Pune")
df
```

```python
def check_alert(df):
    alerts = []

    max_temp = df["temp"].max()
    max_humidity = df["humidity"].max()
    avg_temp = df["temp"].mean()

    if max_temp >= 40:
        alerts.append("🔥 Heat Wave Alert")

    if max_humidity >= 85:
        alerts.append("💧 High Humidity Alert")

    if avg_temp < 25:
        alerts.append("❄️ Cold Weather Alert")

    if len(alerts) == 0:
        alerts.append("✅ Normal Weather Conditions")

    return alerts

alerts = check_alert(df)

print("⚠️ ALERTS:")
for a in alerts:
    print("-", a)
```

```python
plt.figure(figsize=(10,5))

plt.plot(df["datetime"], df["temp"], marker="o", color="red")

plt.xticks(rotation=45)
plt.title("🌡️ Temperature Trend (Simulation)")
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df["datetime"], df["humidity"], marker="s", color="blue")

plt.xticks(rotation=45)
plt.title("💧 Humidity Trend (Simulation)")
plt.xlabel("Time")
plt.ylabel("Humidity (%)")
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.bar(df["datetime"], df["temp"], color="orange")

plt.xticks(rotation=45)
plt.title("🌡️ Hourly Temperature (Bar Chart)")
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.grid()

plt.show()
```
